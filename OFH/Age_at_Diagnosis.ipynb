```r
# Configure population (stored as a dataframe) and predictor here

investigate_PrCa_diagnosis_age <- function(
    population,
    predictor,
    normalise_within_population = FALSE,
    population_name = NA_character_,
    test_name = "GRS vs. Age at PrCa diagnosis",
    make_plot = TRUE,
    print_summary = TRUE,
    print_plot = TRUE
) {

    ## Warnings for improper inputs

    if (!is.data.frame(population)) {
        stop("Error: population must be a dataframe.")
    }

    if (!is.character(predictor) || length(predictor) != 1) {
        stop("Error: predictor must be a single character string.")
    }

    required_outcome <- "age_at_PrCa_diagnosis"

    if (!required_outcome %in% names(population)) {
        stop(
            paste0(
                "Error: outcome column '",
                required_outcome,
                "' is missing from population."
            )
        )
    }

    if (!predictor %in% names(population)) {
        stop(
            paste0(
                "Error: predictor '",
                predictor,
                "' is missing from population."
            )
        )
    }

    ## Keep only relevant columns and complete cases

    analysis_df <- population[, c(required_outcome, predictor)]
    analysis_df <- analysis_df[complete.cases(analysis_df), ]

    case_n <- nrow(analysis_df)

    if (nrow(analysis_df) < 3) {
        stop("Error: fewer than 3 complete observations are available.")
    }

    if (!is.numeric(analysis_df[[predictor]])) {
        stop("Error: predictor must be numeric.")
    }

    ## If normalise_within_population is TRUE, re-scale the predictor variable
    ## to SD units within the specified population

    if (normalise_within_population) {

        predictor_sd <- stats::sd(analysis_df[[predictor]], na.rm = TRUE)

        if (is.na(predictor_sd) || predictor_sd == 0) {
            stop("Error: predictor SD is zero or undefined, so it cannot be normalised.")
        }

        analysis_df[[predictor]] <- as.numeric(scale(analysis_df[[predictor]]))
    }

    ## Build model formula

    model_formula <- reformulate(
        termlabels = predictor,
        response = required_outcome
    )

    ## Linear regression to investigate association

    model <- glm(
        formula = model_formula,
        data = analysis_df,
        family = gaussian()
    )

    model_summary <- summary(model)

    if (print_summary) {
        print(model_summary)
    }

    ## Extract model results

    coef_table <- coef(model_summary)

    intercept <- coef_table["(Intercept)", "Estimate"]

    predictor_coef <- coef_table[predictor, "Estimate"]
    predictor_se <- coef_table[predictor, "Std. Error"]
    predictor_t <- coef_table[predictor, "t value"]
    predictor_p <- coef_table[predictor, "Pr(>|t|)"]

    summary_row <- data.frame(
        Test = test_name,
        Population = population_name,
        Predictor = predictor,
        Normalisation = normalise_within_population,
        `Case N` = case_n,
        `GRS coefficient` = predictor_coef,
        `GRS SE` = predictor_se,
        `GRS t` = predictor_t,
        `GRS p-value` = predictor_p,
        Intercept = intercept,
        check.names = FALSE
    )

    ## Plot association in ggplot2

    plot <- NULL

    if (make_plot) {
        library(ggplot2)

        plot <- ggplot(
            data = analysis_df,
            aes(
                x = .data[[predictor]],
                y = .data[[required_outcome]]
            )
        ) +
            geom_point() +
            geom_smooth(
                method = "glm",
                method.args = list(family = gaussian())
            ) +
            labs(
                x = predictor,
                y = "Age at Prostate Cancer Diagnosis"
            )

        if (print_plot) {
            print(plot)
        }
    }

    ## Return everything useful

    return(
        list(
            model = model,
            model_summary = model_summary,
            plot = plot,
            summary_row = summary_row,
            analysis_data = analysis_df
        )
    )
}


##################################
# Bulk run for multiple analyses #
##################################

bulk_investigate_PrCa_diagnosis_age <- function(
    populations = list(             ## Default list contains all populations, can be overridden by user
        All = PrCa_all_men,
        White = PrCa_white_men,
        Black = PrCa_black_men,
        Mixed = PrCa_mixed_men,
        South_Asian = PrCa_southasian_men,
        Chinese = PrCa_chinese_men,
        Arab = PrCa_arab_men,
        Other = PrCa_other_men,
        EUR = PrCa_EUR_men,
        AFR = PrCa_AFR_men,
        SAS = PrCa_SAS_men,
        EAS = PrCa_EAS_men,
        SEA = PrCa_SEA_men,
        CAS = PrCa_CAS_men,
        AMR = PrCa_AMR_men,
        MID = PrCa_MID_men
    ),
    predictors = c(             ## Default character vector contains all GRSs, can be overridden by user
        ContiMultiethnicGRS = "ContiMultiethnicGRS",
        ContiEURGRS = "ContiEURGRS",
        ContiAFRGRS = "ContiAFRGRS",
        ContiEASGRS = "ContiEASGRS",
        ContiHISGRS = "ContiHISGRS",
        WangMultiethnicGRS = "WangMultiethnicGRS",
        WangEURGRS = "WangEURGRS",
        WangAFRGRS = "WangAFRGRS",
        WangEASGRS = "WangEASGRS",
        WangHISGRS = "WangHISGRS"
    ),
    normalisations = c("Global", "Subpopulation"),  ## Default character vector contains both normalisation options, can be overridden by user
    test_name = "GRS vs. Age at PrCa diagnosis",
    make_plots = FALSE,
    print_summaries = FALSE,
    print_plots = FALSE,
    return_full_results = FALSE
) {

    ## Input checks

    if (!is.list(populations)) {
        stop("Error: populations must be a named list of dataframes.")
    }

    if (is.null(names(populations)) || any(names(populations) == "")) {
        stop("Error: populations must be a named list, e.g. list(EUR = eur_df, AFR = afr_df).")
    }

    if (!all(vapply(populations, is.data.frame, logical(1)))) {
        stop("Error: every object in populations must be a dataframe.")
    }

    if (!is.character(predictors) || length(predictors) < 1) {
        stop("Error: predictors must be a character vector.")
    }

    if (!is.character(normalisations) || length(normalisations) < 1 || any(is.na(normalisations))) {
        stop("Error: normalisations must be a character vector containing 'Global' and/or 'Subpopulation'.")
    }

    normalisations <- unique(normalisations)

    allowed_normalisations <- c("Global", "Subpopulation")

    if (!all(normalisations %in% allowed_normalisations)) {
        stop("Error: normalisations must contain only 'Global' and/or 'Subpopulation'.")
    }

    ## Store outputs

    summary_rows <- list()
    full_results <- list()

    counter <- 1

    ## Run every combination

    for (population_name in names(populations)) {
        for (predictor in predictors) {
            for (normalisation in normalisations) {
                
                normalise_within_population <- normalisation == "Subpopulation"

                result_name <- paste(
                    population_name,
                    predictor,
                    paste0("normalisation_", normalisation),
                    sep = "__"
                )

                result <- tryCatch(

                    investigate_PrCa_diagnosis_age(
                        population = populations[[population_name]],
                        predictor = predictor,
                        normalise_within_population = normalise_within_population,
                        population_name = population_name,
                        test_name = test_name,
                        make_plot = make_plots,
                        print_summary = print_summaries,
                        print_plot = print_plots
                    ),

                    error = function(e) {

                        warning(
                            paste0(
                                "Analysis failed for population '",
                                population_name,
                                "', predictor '",
                                predictor,
                                "', normalisation = ",
                                normalisation,
                                ". Reason: ",
                                e$message
                            )
                        )

                        empty_summary_row <- data.frame(
                            Test = test_name,
                            Population = population_name,
                            Predictor = predictor,
                            Normalisation = normalisation,
                            `Case N` = NA_integer_,
                            `GRS coefficient` = NA_real_,
                            `GRS SE` = NA_real_,
                            `GRS t` = NA_real_,
                            `GRS p-value` = NA_real_,
                            Intercept = NA_real_,
                            check.names = FALSE
                        )

                        list(
                            model = NULL,
                            model_summary = NULL,
                            plot = NULL,
                            summary_row = empty_summary_row,
                            analysis_data = NULL,
                            error = e$message
                        )
                    }
                )

                result$summary_row$Normalisation <- normalisation

                summary_rows[[counter]] <- result$summary_row
                full_results[[result_name]] <- result

                counter <- counter + 1
            }
        }
    }

    ## Bind into one summary table

    summary_table <- do.call(rbind, summary_rows)
    rownames(summary_table) <- NULL

    if (return_full_results) {
        return(
            list(
                summary_table = summary_table,
                full_results = full_results
            )
        )
    } else {
        return(summary_table)
    }
}
```
