```r
## Filter to males only and add age_at_PrCa_diagnosis column

library(dplyr)

all_men_PrCa_GRS <- all_data_PrCa_GRS %>%
    dplyr::filter(demog_sex == 1) %>%
    dplyr::mutate(age_at_PrCa_diagnosis = as.numeric(difftime(earliest_PrCa_date, date_of_birth, units = "days")) / 365.25)

# Scale GRS columns to SD units (z-transform)

grs_prs_cols <- names(all_men_PrCa_GRS)[
  grepl("(GRS|PRS)[0-9]*$", names(all_men_PrCa_GRS)) &
    !grepl("^top10_all_", names(all_men_PrCa_GRS))
]

grs_prs_cols <- grs_prs_cols[sapply(all_men_PrCa_GRS[grs_prs_cols], is.numeric)]

all_men_PrCa_GRS <- all_men_PrCa_GRS %>%
  dplyr::mutate(
    dplyr::across(
      dplyr::all_of(grs_prs_cols),
      ~ as.numeric(scale(.x))
    )
  )
```

```r
# Bonus analysis before proceeding - when z-transformed to the population as a whole, do Black men have a higher GRS than White men?

all_men_PrCa_GRS %>%
    dplyr::filter(ethnicity_group_narrow == "White") %>%
    dplyr::summarise(
        "Mean Conti GRS (White Men)" = mean(ContiMultiethnicGRS, na.rm = TRUE),
        "Mean Wang GRS (White Men)" = mean(WangMultiethnicGRS, na.rm = TRUE)
    )

all_men_PrCa_GRS %>%
    dplyr::filter(ethnicity_group_narrow == "Black") %>%
    dplyr::summarise(
        "Mean Conti GRS (Black Men)" = mean(ContiMultiethnicGRS, na.rm = TRUE),
        "Mean Wang GRS (Black Men)" = mean(WangMultiethnicGRS, na.rm = TRUE)
    )

## Visualise as a density plot

library(ggplot2)

ggplot(data = all_men_PrCa_GRS %>% dplyr::filter(ethnicity_group_narrow == "White" | ethnicity_group_narrow == "Black"), 
       aes(x = ContiMultiethnicGRS, colour = as.factor(ethnicity_group_narrow))) +
        geom_density() +
        labs(title = "Conti GRS") +
        theme_bw()

ggplot(data = all_men_PrCa_GRS %>% dplyr::filter(ethnicity_group_narrow == "White" | ethnicity_group_narrow == "Black"), 
       aes(x = WangMultiethnicGRS, colour = as.factor(ethnicity_group_narrow))) +
        geom_density() +
        labs(title = "Wang GRS") +
        theme_bw()
```

```r
## Filter to prostate cancer cases only (since we're only interested in those who have been diagnosed)
## Also filter out HES-only and death-only cases, as well as those with no linked genetic data

all_men_PrCa_cases_only <- all_men_PrCa_GRS %>%
    dplyr::filter(!is.na(earliest_PrCa_date)) %>%
                  dplyr::mutate(
                    HES_only = PrCa == 1 & ((is.na(admidate_cancerregistry) & is.na(admidate_death)) & (!is.na(admidate_outpat) | !is.na(admidate_inpat))),
                    death_only = PrCa == 1 & (is.na(admidate_outpat) & is.na(admidate_inpat) & is.na(admidate_cancerregistry) & !is.na(admidate_death))
                  ) %>%
    dplyr::filter(HES_only == FALSE & death_only == FALSE & !is.na(ContiMultiethnicGRS))
```

```r
## Subset to ethnicity/genetic similarity groups

PrCa_all_men <- all_men_PrCa_cases_only

PrCa_white_men <- PrCa_all_men %>%
    dplyr::filter(ethnicity_group_narrow == "White")

PrCa_black_men <- PrCa_all_men %>%
    dplyr::filter(ethnicity_group_narrow == "Black")

PrCa_mixed_men <- PrCa_all_men %>%
    dplyr::filter(ethnicity_group_narrow == "Mixed White and Black")

PrCa_southasian_men <- PrCa_all_men %>%
    dplyr::filter(ethnicity_group_narrow == "South Asian")

PrCa_chinese_men <- PrCa_all_men %>%
    dplyr::filter(ethnicity_group_narrow == "Chinese")

PrCa_arab_men <- PrCa_all_men %>%
    dplyr::filter(ethnicity_group_narrow == "Arab")

PrCa_other_men <- PrCa_all_men %>%
    dplyr::filter(ethnicity_group_narrow == "Other")


# Create separate dataframes for genetic similarity groups

PrCa_EUR_men <- PrCa_all_men %>%
    dplyr::filter(genetic_similarity_group == "EUR")

PrCa_AFR_men <- PrCa_all_men %>%
    dplyr::filter(genetic_similarity_group == "AFR")

PrCa_SAS_men <- PrCa_all_men %>%
    dplyr::filter(genetic_similarity_group == "SAS")

PrCa_EAS_men <- PrCa_all_men %>%
    dplyr::filter(genetic_similarity_group == "EAS")

PrCa_SEA_men <- PrCa_all_men %>%
    dplyr::filter(genetic_similarity_group == "SEA")

PrCa_CAS_men <- PrCa_all_men %>%
    dplyr::filter(genetic_similarity_group == "CAS")

PrCa_AMR_men <- PrCa_all_men %>%
    dplyr::filter(genetic_similarity_group == "AMR")

PrCa_MID_men <- PrCa_all_men %>%
    dplyr::filter(genetic_similarity_group == "MID")

## write all to csv to take a look

write.csv(PrCa_all_men, "PrCa_all_men.csv", row.names = FALSE)
write.csv(PrCa_white_men, "PrCa_white_men.csv", row.names = FALSE)
write.csv(PrCa_black_men, "PrCa_black_men.csv", row.names = FALSE)
write.csv(PrCa_mixed_men, "PrCa_mixed_men.csv", row.names = FALSE)
write.csv(PrCa_southasian_men, "PrCa_southasian_men.csv", row.names = FALSE)
write.csv(PrCa_chinese_men, "PrCa_chinese_men.csv", row.names = FALSE)
write.csv(PrCa_arab_men, "PrCa_arab_men.csv", row.names = FALSE)
write.csv(PrCa_other_men, "PrCa_other_men.csv", row.names = FALSE)
write.csv(PrCa_EUR_men, "PrCa_EUR_men.csv", row.names = FALSE)
write.csv(PrCa_AFR_men, "PrCa_AFR_men.csv", row.names = FALSE)
write.csv(PrCa_SAS_men, "PrCa_SAS_men.csv", row.names = FALSE)
write.csv(PrCa_EAS_men, "PrCa_EAS_men.csv", row.names = FALSE)
write.csv(PrCa_SEA_men, "PrCa_SEA_men.csv", row.names = FALSE)
write.csv(PrCa_CAS_men, "PrCa_CAS_men.csv", row.names = FALSE)
write.csv(PrCa_AMR_men, "PrCa_AMR_men.csv", row.names = FALSE)
write.csv(PrCa_MID_men, "PrCa_MID_men.csv", row.names = FALSE)

## Counts

print(paste("Number of prostate cancer cases (all men):", nrow(PrCa_all_men)))
print(paste("Number of prostate cancer cases (white men):", nrow(PrCa_white_men)))
print(paste("Number of prostate cancer cases (black men):", nrow(PrCa_black_men)))
print(paste("Number of prostate cancer cases (mixed men):", nrow(PrCa_mixed_men)))
print(paste("Number of prostate cancer cases (south asian men):", nrow(PrCa_southasian_men)))
print(paste("Number of prostate cancer cases (chinese men):", nrow(PrCa_chinese_men)))
print(paste("Number of prostate cancer cases (arab men):", nrow(PrCa_arab_men)))
print(paste("Number of prostate cancer cases (other men):", nrow(PrCa_other_men)))
print(paste("Number of prostate cancer cases (EUR men):", nrow(PrCa_EUR_men)))
print(paste("Number of prostate cancer cases (AFR men):", nrow(PrCa_AFR_men)))
print(paste("Number of prostate cancer cases (SAS men):", nrow(PrCa_SAS_men)))
print(paste("Number of prostate cancer cases (EAS men):", nrow(PrCa_EAS_men)))
print(paste("Number of prostate cancer cases (SEA men):", nrow(PrCa_SEA_men)))
print(paste("Number of prostate cancer cases (CAS men):", nrow(PrCa_CAS_men)))
print(paste("Number of prostate cancer cases (AMR men):", nrow(PrCa_AMR_men)))
print(paste("Number of prostate cancer cases (MID men):", nrow(PrCa_MID_men)))
```

```r
# Configure population (stored as a dataframe) and predictor here

population <- PrCa_all_men          # Change "PrCa_all_men" to the desired population (e.g., "PrCa_white_men", "PrCa_black_men", etc.) to analyze different populations
predictor <- "ContiMultiethnicGRS"  # Change "ContiMultiethnicGRS" to the desired predictor variable name (e.g., "ContiEURGRS", "WangAFRGRS", etc.) to analyze different predictors

## Build model formula
formula <- reformulate(
    termlabels = predictor,
    response = "age_at_PrCa_diagnosis"
)

## Linear regression to investigate association

model <- glm(formula = formula, data = population, family = gaussian())
summary(model)

## Plot association in ggplot2

library(ggplot2)

ggplot(data = population, aes(x = .data[[predictor]], y = .data[["age_at_PrCa_diagnosis"]])) +
    geom_point() +
    geom_smooth(method = "glm", method.args = list(family = gaussian())) +
    labs(x = predictor, y = "Age at Prostate Cancer Diagnosis")
```
