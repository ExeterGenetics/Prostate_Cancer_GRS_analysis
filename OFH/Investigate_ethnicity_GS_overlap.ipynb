## Investigating the overlap between self-reported ethnicity and genetic similarity

This notebook takes the all_data_PrCa.csv file from part 1 of Logistic_Regressions.ipynb, then investigates the overlap between self-reported ethnicity and genetic similarity groupings

If you do not already have the all_data_PrCa.csv file in your project space, you will need to run Logistic_Regressions.ipynb up to the end of Part 1 first.

```r
## Download the dataset from DNAnexus

system("bash -c 'unset HTTP_PROXY HTTPS_PROXY http_proxy https_proxy; dx download project-J8PJg1V25y16J0Y3F05pPJBj:/users/callum/Derived_datasets/all_data_PrCa.csv'")

## load in the dataset

all_data_PrCa <- read.csv("all_data_PrCa.csv")

## Subset to those with a genetic similarity call

gs_vs_ethnicity <- all_data_PrCa %>%
    filter(!is.na(genetic_similarity_group))
```

```r
print("By genetic similarity group, what do people self-report?")

library(dplyr)

ethnicity_by_gs <- gs_vs_ethnicity %>%
    group_by(genetic_similarity_group, ethnicity_group_narrow) %>%
    summarise(n = n(), .groups = "drop")



ethnicity_by_gs

print("By ethnicity group, what do people cluster as in genetic similarity groups?")

library(dplyr)

gs_by_ethnicity <- gs_vs_ethnicity %>%
    group_by(ethnicity_group_narrow, genetic_similarity_group) %>%
    summarise(n = n(), .groups = "drop")



gs_by_ethnicity
```

```r
## Overlap summary table

library(dplyr)

overlap_summary_table <- gs_vs_ethnicity %>%
  summarise(
    `N (Black)` = sum(
      ethnicity_group_narrow == "Black",
      na.rm = TRUE
    ),
    `N (AFR)` = sum(
      genetic_similarity_group == "AFR",
      na.rm = TRUE
    ),
    `N (Black & AFR)` = sum(
      ethnicity_group_narrow == "Black" &
        genetic_similarity_group == "AFR",
      na.rm = TRUE
    ),
    `N (Black | AFR)` = sum(
      ethnicity_group_narrow == "Black" |
        genetic_similarity_group == "AFR",
      na.rm = TRUE
    ),
    `Overlap Black + AFR %` = round(
      100 * `N (Black & AFR)` / `N (Black | AFR)`,
      1
    ),
    `N (White)` = sum(
      ethnicity_group_narrow == "White",
      na.rm = TRUE
    ),
    `N (EUR)` = sum(
      genetic_similarity_group == "EUR",
      na.rm = TRUE
    ),
    `N (White & EUR)` = sum(
      ethnicity_group_narrow == "White" &
        genetic_similarity_group == "EUR",
      na.rm = TRUE
    ),
    `N (White | EUR)` = sum(
      ethnicity_group_narrow == "White" |
        genetic_similarity_group == "EUR",
      na.rm = TRUE
    ),
    `Overlap White & EUR %` = round(
      100 * `N (White & EUR)` / `N (White | EUR)`,
      1
    )
  )

overlap_summary_table
```
