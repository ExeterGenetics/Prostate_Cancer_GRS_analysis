```python
# This codeblock joins the Prostate cancer cases to the baseline data, which includes all participants

## call in baseline data

baseline_spark <- dplyr::tbl(sc, dbplyr::in_schema(db, "participant"))
baseline_all <- baseline_spark |> dplyr::collect()
write.csv(baseline_all, "baseline_all.csv", row.names = FALSE)

## Fetch list of participants who actually have linked healthcare records

linked_spark <- dplyr::tbl(sc, dbplyr::in_schema(db, "participant_nhs_linked"))
linked <- linked_spark |> dplyr::collect()

linked <- linked %>%
    dplyr::select("pid") 

## Left join baseline onto linked (to ensure that no one without linked healthcare records is included)

baseline <- merge(linked, baseline_all, by = "pid", all.x = TRUE)
write.csv(baseline, "baseline.csv", row.names = FALSE)

## Merge cleaned baseline data with PrCa diagnosis data

all_data <- merge(baseline, PrCa_earliest, by = "pid", all.x = TRUE) %>% ## left join
    dplyr::mutate(PrCa = ifelse(is.na(PrCa), 0, PrCa)) # if they don't have prostate cancer, they'll be NA by default. Let's set these to 0

all_data <- coalesce_pairs(all_data) # combine the gender and sex columns into a single column
```

```python
## This codeblock then excludes controls with any of the control exclusion codes written at the top of the script

exclusions_outpat <- read_ICD10(ICD10_control_exclusion_codes, table = "nhse_eng_outpat", baseline = FALSE, all_cols = FALSE) %>%
  dplyr::select("pid")

exclusions_inpat <- read_ICD10(ICD10_control_exclusion_codes, table = "nhse_eng_inpat", baseline = FALSE, all_cols = FALSE) %>%
  dplyr::select("pid")

exclusions_cancerregistry <- read_ICD10(ICD10_control_exclusion_codes, table = "nhse_eng_canreg_pattumour", baseline = FALSE, all_cols = FALSE) %>%
  dplyr::select("pid")

exclusions_death <- read_ICD10(ICD10_control_exclusion_codes, table = "nhse_engwal_deaths", baseline = FALSE, all_cols = FALSE) %>%
  dplyr::select("pid")

exclusions_OPCS <- read_OPCS(OPCS_control_exclusion_codes) %>%
  dplyr::select("pid")

possible_PrCa_cases <- merge(exclusions_outpat, exclusions_inpat, by = "pid", all = T) %>%
    merge(exclusions_cancerregistry, by = "pid", all = T) %>%
    merge(exclusions_death, by = "pid", all = T) %>%
    merge(exclusions_OPCS, by = "pid", all = T)

## collapse exclusions to one row per participant, leaving only pids and a binary variable for whether they meet any of the possible PrCa exclusion criteria 

possible_PrCa_cases <- possible_PrCa_cases %>%
  group_by(pid) %>%
  dplyr::summarise(possible_PrCa_case = 1)
```

```python
## This codeblock joins possible_PrCa_cases to the main dataframe and excludes controls

all_data_exclusions <- merge(all_data, possible_PrCa_cases, by = "pid", all.x = TRUE) %>%
  dplyr::mutate(exclude = if_else(possible_PrCa_case == 1 & (PrCa == 0 | is.na(PrCa)), 1L, 0L, missing = 0L))

all_data_exclusions <- all_data_exclusions %>%
    dplyr::filter(exclude != 1)  # exclude participants who meet any of the possible PrCa case criteria but don't have a confirmed PrCa diagnosis 

## Create a dataframe of pid, PrCa status, earliest date of diagnosis, and source of earliest date of diagnosis

PrCa_cases_dates <- all_data_exclusions %>% 
    dplyr::select("pid", "PrCa", "earliest_PrCa_date", "earliest_date_source")

write.csv(PrCa_cases_dates, "PrCa_cases_dates.csv", row.names = FALSE)

## Write to csv

write.csv(all_data_exclusions, "all_data_exclusions.csv", row.names = FALSE)
```

```python
# Add extra variables dob, registration_date, age_at_registration, and Male (TRUE/FALSE) to all_data

all_data_plus_extra_vars <- all_data_exclusions

# 1. Set all dates of birth to 16th of the month

all_data_plus_extra_vars$date_of_birth <- as.Date(
    sprintf("%04d-%02d-16", all_data_plus_extra_vars$birth_year, all_data_plus_extra_vars$birth_month),
            format = "%Y-%m-%d"
)

# 2. Set registration dates to the 16th of the month

all_data_plus_extra_vars$registration_date <- as.Date(
    sprintf("%04d-%02d-16", all_data_plus_extra_vars$registration_year, all_data_plus_extra_vars$registration_month),
            format = "%Y-%m-%d"
)   

# 3. Create "date40", "date45", "date50", "date55", "date60", "date65" as the dates at which the participants turned 40, 45, 50, 55, 60, and 65 years old (to be used as timepoints for age-specific analyses)

library(lubridate)

all_data_plus_extra_vars <- all_data_plus_extra_vars %>%
    dplyr::mutate(
        date40 = date_of_birth + lubridate::years(40),
        date45 = date_of_birth + lubridate::years(45),
        date50 = date_of_birth + lubridate::years(50),
        date55 = date_of_birth + lubridate::years(55),
        date60 = date_of_birth + lubridate::years(60),
        date65 = date_of_birth + lubridate::years(65)
    )

# 4. Set age_at_registration to the difference between registration_date and date_of_birth

all_data_plus_extra_vars$age_at_registration <- as.numeric(difftime(all_data_plus_extra_vars$registration_date, all_data_plus_extra_vars$date_of_birth, units = "days")) / 365.25

# 5. Create a "Male" (TRUE/FALSE) and a "Female" (TRUE/FALSE) column. Females will be filtered out later

all_data_plus_extra_vars <- all_data_plus_extra_vars %>%
    dplyr::mutate(
        male = ifelse(demog_sex == 1, TRUE, FALSE),
        female = ifelse(demog_sex == 2, TRUE, FALSE)
    )

# 6. Create "ethnicity_group_wide" column (where mixed participants are classes according to their minority ethnic group)

all_data_plus_extra_vars <- all_data_plus_extra_vars %>%
    dplyr::mutate(
        ethnicity_group_wide = case_when(
            demog_ethnicity_1_1 %in% c(1,  # 1 = White - English / Welsh / Scottish / Northern Irish / British
                                       2,  # 2 = White Irish
                                       3,  # 3 = White Gypsy or Irish Traveller
                                       4,  # 4 = White Polish
                                       5)  # 5 = Any other White background
                                         ~ "White",
            demog_ethnicity_1_1 %in% c(6,  # 6 = Mixed White and Black Caribbean
                                       7,  # 7 = Mixed White and Black African
                                       15, # 15 = Black or Black British - African
                                       16, # 16 = Black or Black British - Caribbean
                                       17) # 17 = Any other Black / African / Caribbean background
                                         ~ "Black",
            demog_ethnicity_1_1 %in% c(10, # 10 = Asian or Asian British - Indian
                                       11, # 11 = Asian or Asian British - Pakistani
                                       12) # 12 = Asian or Asian British - Bangladeshi
                                         ~ "South Asian",
            demog_ethnicity_1_1 %in% c(8,  # 8 = Mixed White and Asian
                                       13, # 13 = Chinese
                                       14) # 14 = Any other Asian/Asian British background
                                         ~ "East Asian",
            demog_ethnicity_1_1 %in% c(18) ~ "Arab",
            demog_ethnicity_1_1 %in% c(9,  # 9 = Any other mixed multiple ethnic background
                                       19, # 19 = Other
                                       -3) # -3 = Prefer not to answer
                                        ~ "Other",
    TRUE ~ NA_character_
  ))

# 7. Create "ethnicity_group_narrow" column (where mixed ethnicity participants are categorised separately)

all_data_plus_extra_vars <- all_data_plus_extra_vars %>%
    dplyr::mutate(ethnicity_group_narrow = case_when(
        demog_ethnicity_1_1 %in% c(1,  # 1 = White - English / Welsh / Scottish / Northern Irish / British
                                   2,  # 2 = White Irish
                                   3,  # 3 = White Gypsy or Irish Traveller
                                   4,  # 4 = White Polish
                                   5)  # 5 = Any other White background
                                     ~ "White",
        demog_ethnicity_1_1 %in% c(6,  # 6 = Mixed White and Black Caribbean
                                   7)  # 7 = Mixed White and Black African
                                     ~ "Mixed White and Black",
        demog_ethnicity_1_1 %in% c(15, # 15 = Black or Black British - African
                                   16, # 16 = Black or Black British - Caribbean
                                   17) # 17 = Any other Black / African / Caribbean background
                                     ~ "Black",
        demog_ethnicity_1_1 %in% c(10, # 10 = Asian or Asian British - Indian
                                   11, # 11 = Asian or Asian British - Pakistani
                                   12) # 12 = Asian or Asian British - Bangladeshi
                                     ~ "South Asian",
        demog_ethnicity_1_1 %in% c(8) ~ "Mixed White and Asian",
        demog_ethnicity_1_1 %in% c(13) ~ "Chinese",
        demog_ethnicity_1_1 %in% c(18) ~ "Arab",
        demog_ethnicity_1_1 %in% c(9,  # 9 = Any other mixed multiple ethnic background
                                   14, # 14 = Any other Asian/Asian British background
                                   19, # 19 = Other
                                   -3) # -3 = Prefer not to answer
                                     ~ "Other",
        TRUE ~ NA_character_
    ))

# 8. Add "Genetic ancestry" (genetic similarity) column 

system(paste("dx download", "imputed_resources/ofh_imputed_sample_ancestry.v6.tsv")) 
genetic_similarity <- read.table("ofh_imputed_sample_ancestry.v6.tsv", header = TRUE, sep = "\t") %>%
    dplyr::select("id", "ANCESTRY_CALL") %>%
    dplyr::rename("pid" = "id", "genetic_similarity" = "ANCESTRY_CALL") %>%
    dplyr::mutate(genetic_similarity_group = case_when(
        genetic_similarity %in% c("CW_EUROPE", "C_S_UK", "E_EUROPE", "FINLAND", "IRELAND", "NI_N_SCOT",
                                  "NW_WALES", "N_ENG_S_SCOT", "N_EUROPE", "PORTUGAL_SPAIN", "SE_EUROPE",
                                  "SW_WALES") ~ "EUR",
        genetic_similarity %in% c("C_S_AFRICA", "E_AFRICA", "N_AFRICA", "W_AFRICA") ~ "AFR",
        genetic_similarity %in% c("BANGLADESH", "INDIA_PAKISTAN", "SRI_LANKA") ~ "SAS",
        genetic_similarity %in% c("CE_ASIA", "JAPAN_KOREA") ~ "EAS",
        genetic_similarity %in% c("SE_ASIA") ~ "SEA",
        genetic_similarity %in% c("C_ASIA") ~ "CAS",
        genetic_similarity %in% c("N_C_S_AMERICA") ~ "AMR",
        genetic_similarity %in% c("M_EAST_W_ASIA") ~ "MID",
        TRUE ~ NA_character_
    )
    )

all_data_plus_extra_vars <- merge(all_data_plus_extra_vars, genetic_similarity %>% dplyr::select(pid, genetic_similarity, genetic_similarity_group), by = "pid", all.x = TRUE)

# Checkpoint, since running to this point takes around 30 minutes - dx upload all_data_plus_extra_vars.csv:

all_data_PrCa <- all_data_plus_extra_vars

write.csv(all_data_PrCa, "all_data_PrCa.csv", row.names = FALSE)
system(paste("dx upload", "all_data_PrCa.csv"))

print("Ensure you move all_data_PrCa.csv to /Derived_datasets in the GOCT project, or a folder of your choice. Also ensure it is the only file with this name in the folder")
```

```python
# Tag participants who already had prostate cancer at a chosen "screening date" as "pre_diagnosed". These participants will be filtered out later
# Any death-only PrCa cases will be treated as pre_diagnosed, as we don't have HES or Cancer Registry dates to prove that they were diagnosed after
# the screening date.
# (furthermore, death records are more up-to-date than cancer registry records, so in most cases there is simply a delay in the release of the cancer registry dates)

# Set the screening date here, to any of the following:
#    - registration_date
#    - date40
#    - date45
#    - date50
#    - date55
#    - date60
#    - date65

## Ensure date columns are in date format
date_columns <- c("date_of_birth", "registration_date", "date40", "date45", "date50", "date55", "date60", "date65",
                  "admidate_outpat", "admidate_inpat", "admidate_cancerregistry", "admidate_death", "earliest_PrCa_date")

all_data_PrCa_GRS[date_columns] <- lapply(all_data_PrCa_GRS[date_columns], as.Date)

## Set screening date

screening_date <- "date50"
current_date <- as.Date(sub(".*study_ofhs260023_(\\d{8}).*", "\\1", db), format = "%Y%m%d")
print(paste("Current date (from database name):", current_date))

## Define "pre_diagnosed"

PrCa_prediagnosis <- all_data_PrCa_GRS %>%
    dplyr::mutate(
        pre_diagnosed = PrCa == 1 & (                                                                                       # pre_diagnosed means any of the following...
            (is.na(admidate_outpat) & is.na(admidate_inpat) & is.na(admidate_cancerregistry) & is.na(admidate_death)) |     # if they have no dates of diagnosis (since we can't be sure when they were diagnosed)
            (!is.na(admidate_outpat) & admidate_outpat < .data[[screening_date]]) |                                         # if they have an outpatient diagnosis date before the screening date    
            (!is.na(admidate_inpat) & admidate_inpat < .data[[screening_date]]) |                                           # if they have an inpatient diagnosis date before the screening date
            (!is.na(admidate_cancerregistry) & admidate_cancerregistry < .data[[screening_date]]) |                         # if they have a cancer registry diagnosis date before the screening date
            (!is.na(admidate_death) & admidate_death < .data[[screening_date]])                                             # if they have a death date before the screening date
        ),
       earliest_PrCa_date = pmin(admidate_outpat, admidate_inpat, admidate_cancerregistry, admidate_death, na.rm = TRUE) 
    )

## read in dates of death

death_registry_spark <- dplyr::tbl(sc, dbplyr::in_schema(db, "nhse_engwal_deaths"))
death_registry <- death_registry_spark |> dplyr::collect()
write.csv(death_registry, "death_registry.csv", row.names = FALSE)

death_dates <- death_registry %>%
    dplyr::select(pid, reg_date_of_death)

## Tag participants who had a prostate cancer record in inpatient and/or outpatient data but not in cancer registry or death records as "HES-only" (these will be filtered out later)
# also tag participants with prostate cancer record in death records but not in inpatient, outpatient, or cancer registry data as "death-only" (these will also be filtered out later)
# also add "current age" and "old_enough" flag - if current age is less than the screening age + 2, 5, or 10 years, then old enough = FALSE
# also add "died_within_2yrs", "died_within_5yrs", and "died_within_10yrs" flags - if they died within 2, 5, or 10 years of the screening date, then these flags = TRUE

PrCa_prediagnosis <- PrCa_prediagnosis %>%
    dplyr::mutate(
        HES_only = PrCa == 1 & ((is.na(admidate_cancerregistry) & is.na(admidate_death)) & (!is.na(admidate_outpat) | !is.na(admidate_inpat))),
        death_only = PrCa == 1 & (is.na(admidate_outpat) & is.na(admidate_inpat) & is.na(admidate_cancerregistry) & !is.na(admidate_death)),
        current_age = as.numeric(difftime(current_date, date_of_birth, units = "days")) / 365.25,
        old_enough_screening_age = current_age >= as.numeric(difftime(.data[[screening_date]], date_of_birth, units = "days")) / 365.25,
        old_enough_2yrs = current_age >= as.numeric(difftime(.data[[screening_date]], date_of_birth, units = "days")) / 365.25 + 2,
        old_enough_5yrs = current_age >= as.numeric(difftime(.data[[screening_date]], date_of_birth, units = "days")) / 365.25 + 5,
        old_enough_10yrs = current_age >= as.numeric(difftime(.data[[screening_date]], date_of_birth, units = "days")) / 365.25 + 10
    ) %>%
    dplyr::left_join(death_dates, by = "pid") %>%
    dplyr::mutate(
        died_within_2yrs = !is.na(reg_date_of_death) & (reg_date_of_death <= .data[[screening_date]] + lubridate::years(2)),
        died_within_5yrs = !is.na(reg_date_of_death) & (reg_date_of_death <= .data[[screening_date]] + lubridate::years(5)),
        died_within_10yrs = !is.na(reg_date_of_death) & (reg_date_of_death <= .data[[screening_date]] + lubridate::years(10))
    )

write.csv(PrCa_prediagnosis, "PrCa_prediagnosis.csv", row.names = FALSE)
```

```python
## Set up logistic regression and confusion matrix function

## Subset controls to match the number of cases

subset_controls_to_case_count <- function(df, case_col) {
  cases    <- df[!is.na(df[[case_col]]) & df[[case_col]] == 1L, , drop = FALSE]
  controls <- df[!is.na(df[[case_col]]) & df[[case_col]] == 0L, , drop = FALSE]

  ncases    <- nrow(cases)
  ncontrols <- nrow(controls)

  if (ncases == 0L || ncontrols == 0L) {
    return(df)
  }

  if (ncontrols < ncases) {
    warning("Fewer controls than cases; returning all controls and all cases.")
  }

  n_to_sample     <- min(ncases, ncontrols)
  sampled_controls <- controls[sample.int(ncontrols, size = n_to_sample, replace = FALSE), , drop = FALSE]

  dplyr::bind_rows(cases, sampled_controls)
}

## Run logistic regressions + ROC curve

run_logreg <- function(data,
                       outcome,
                       predictor,
                       covariates = NULL,
                       subset_controls = FALSE,
                       plot_roc = TRUE,
                       plot_pr = FALSE,
                       show_output = TRUE,
                       remove_dead_controls = TRUE,
                       remove_too_young = TRUE) {

  if (isTRUE(subset_controls)) {
    data <- subset_controls_to_case_count(data, case_col = outcome)
  }

  if (isTRUE(remove_dead_controls)) {
      if(outcome == "PrCa_2yrs") {
          data <- data %>% dplyr::filter((died_within_2yrs == FALSE & PrCa_2yrs != 1) | PrCa_2yrs == 1)
      } else if(outcome == "PrCa_5yrs") {
          data <- data %>% dplyr::filter((died_within_5yrs == FALSE & PrCa_5yrs != 1) | PrCa_5yrs == 1)
      } else if(outcome == "PrCa_10yrs") {
          data <- data %>% dplyr::filter((died_within_10yrs == FALSE & PrCa_10yrs != 1) | PrCa_10yrs == 1)
      } else {
          warning("Outcome not recognised for remove_dead; no participants removed based on death within prediction horizon.")
      }
  }

  if (isTRUE(remove_too_young)) {
      if(outcome == "PrCa" | outcome == "PrCa_post_assessment") {
          data <- data %>% dplyr::filter(old_enough_screening_age == TRUE)
      } else if(outcome == "PrCa_2yrs") {
          data <- data %>% dplyr::filter(old_enough_2yrs == TRUE)
      } else if(outcome == "PrCa_5yrs") {
          data <- data %>% dplyr::filter(old_enough_5yrs == TRUE)
      } else if(outcome == "PrCa_10yrs") {
          data <- data %>% dplyr::filter(old_enough_10yrs == TRUE)
      } else {
          warning("Outcome not recognised for remove_too_young; no participants removed based on age within prediction horizon.")
      }
  }

  if (is.null(covariates)) {
    formula <- as.formula(paste(outcome, "~", predictor))
  } else {
    formula <- as.formula(
      paste(outcome, "~", predictor, "+", paste(covariates, collapse = " + "))
    )
  }
  
  logreg <- glm(formula, data = data, family = binomial)
  if (show_output) {
    print(summary(logreg))
  }
  data$pred <- predict(logreg, data, type = "response")
  data$predtop10 <- data$pred >= quantile(data$pred, probs = 0.9, na.rm = TRUE, names = FALSE)
  data$predtop20 <- data$pred >= quantile(data$pred, probs = 0.8, na.rm = TRUE, names = FALSE)
  data$predtop30 <- data$pred >= quantile(data$pred, probs = 0.7, na.rm = TRUE, names = FALSE)
  data$predtop40 <- data$pred >= quantile(data$pred, probs = 0.6, na.rm = TRUE, names = FALSE)
  data$predtop50 <- data$pred >= quantile(data$pred, probs = 0.5, na.rm = TRUE, names = FALSE)
  data$predtop60 <- data$pred >= quantile(data$pred, probs = 0.4, na.rm = TRUE, names = FALSE)
  data$predtop70 <- data$pred >= quantile(data$pred, probs = 0.3, na.rm = TRUE, names = FALSE)
  data$predtop80 <- data$pred >= quantile(data$pred, probs = 0.2, na.rm = TRUE, names = FALSE)
  data$predtop90 <- data$pred >= quantile(data$pred, probs = 0.1, na.rm = TRUE, names = FALSE)

  proc_available <- tryCatch(
    requireNamespace("pROC", quietly = TRUE),
    error = function(e) FALSE
  )

  roc_obj <- NULL
  if (!proc_available) {
    warning("pROC is unavailable; ROC skipped. Use the manual ROC cell below.")
  } else {
    if (plot_roc && show_output) {
      roc_obj <- roc(data[[outcome]] ~ data$pred,
                     plot = TRUE,
                     print.auc = TRUE,
                     ci = TRUE)
    } else {
      roc_obj <- roc(data[[outcome]] ~ data$pred,
                     ci = TRUE)
    }
    if (show_output) {
      print(roc_obj)
    }
  }
  
  pr_obj <- NULL
  if (plot_pr) {
    labels <- as.numeric(data[[outcome]] == 1)
    scores <- data$pred
    keep <- !is.na(labels) & !is.na(scores)
    labels_clean <- labels[keep]
    scores_clean <- scores[keep]
    prevalence <- mean(labels_clean == 1)

    prroc_available <- tryCatch(
      requireNamespace("PRROC", quietly = TRUE),
      error = function(e) FALSE
    )

    if (!prroc_available) {
      warning("PR curve skipped: PRROC is unavailable or failed to load in this session.")
    } else {
      pr_obj <- tryCatch(
        PRROC::pr.curve(
          scores.class0 = scores_clean[labels_clean == 1],
          scores.class1 = scores_clean[labels_clean == 0],
          curve = TRUE
        ),
        error = function(e) {
          warning(paste0("PR curve skipped due to PRROC error: ", e$message))
          NULL
        }
      )

      if (!is.null(pr_obj) && plot_pr && show_output) {
        model_label <- if (is.null(covariates)) {
          predictor
        } else {
          paste(predictor, "+", paste(covariates, collapse = " + "))
        }
        plot(pr_obj, main = paste("Precision-Recall Curve:", model_label),
             xlab = "Recall", ylab = "Precision")
        abline(h = prevalence, col = "red", lty = 2, lwd = 2)
        legend("bottomleft",
               legend = paste0("Outcome prevalence = ", sprintf("%.3f", prevalence)),
               col = "red", lty = 2, lwd = 2, bty = "n")
      }
    }
    
    if (show_output) {
      print(pr_obj)
    }
  }
  
  return(list(
    model = logreg,
    data = data,
    outcome = outcome,
    predictor = predictor,
    covariates = covariates,
    roc = roc_obj,
    pr = pr_obj
  ))
}

## Confusion matrix

confusion_matrix <- function(data,
                             outcome,
                             cutoff_value = 0.5,
                             positive_level = 1,
                             negative_level = 0,
                             print_epi = TRUE) {

  if (cutoff_value <= 0 || cutoff_value >= 1) {
    stop("cutoff_value must be between 0 and 1 (exclusive).")
  }
  if (!("pred" %in% names(data))) {
    stop("data must contain a column named 'pred' with predicted probabilities.")
  }

  # Predicted class at threshold (treat/positive if pred >= pt)
  data$predbin <- factor(
    data$pred >= cutoff_value,
    levels = c(TRUE, FALSE)
  )

  # Actual outcome (ensure positive then negative ordering)
  data$outcome_actual <- factor(
    data[[outcome]],
    levels = c(positive_level, negative_level)
  )

  # Confusion matrix: rows = predicted (TRUE/FALSE), cols = actual (pos/neg)
  cm <- table(
    pred_outcome = data$predbin,
    outcome_actual = data$outcome_actual
  )

  print(cm)

  get_cell <- function(mat, r, c) {
    r <- as.character(r); c <- as.character(c)
    if (r %in% rownames(mat) && c %in% colnames(mat)) mat[r, c] else 0
  }

  TP <- get_cell(cm, TRUE,  positive_level)
  FP <- get_cell(cm, TRUE,  negative_level)
  FN <- get_cell(cm, FALSE, positive_level)
  TN <- get_cell(cm, FALSE, negative_level)
  N  <- sum(cm)

  epi_available <- tryCatch(
    requireNamespace("epiR", quietly = TRUE),
    error = function(e) FALSE
  )

  if (epi_available) {
    epi <- epiR::epi.tests(cm)
    if (print_epi) print(epi)
  } else {
    warning("epi.tests is unavailable; returning manual diagnostic metrics.")

    safe_div <- function(num, den) ifelse(den == 0, NA_real_, num / den)

    sensitivity <- safe_div(TP, TP + FN)
    specificity <- safe_div(TN, TN + FP)
    ppv <- safe_div(TP, TP + FP)
    npv <- safe_div(TN, TN + FN)
    accuracy <- safe_div(TP + TN, N)
    f1_score <- safe_div(2 * ppv * sensitivity, ppv + sensitivity)
    lr_pos <- safe_div(sensitivity, 1 - specificity)
    lr_neg <- safe_div(1 - sensitivity, specificity)

    epi <- list(
      method = "manual",
      confusion_matrix = cm,
      counts = list(TP = TP, FP = FP, FN = FN, TN = TN, N = N),
      metrics = list(
        sensitivity = sensitivity,
        specificity = specificity,
        ppv = ppv,
        npv = npv,
        accuracy = accuracy,
        f1_score = f1_score,
        lr_pos = lr_pos,
        lr_neg = lr_neg
      )
    )

    if (print_epi) {
      cat("\nManual diagnostic metrics (epi.tests fallback):\n")
      print(epi$counts)
      print(epi$metrics)
    }
  }

  # ---- Net benefit calculations ----
  pt <- cutoff_value
  w  <- pt / (1 - pt)  # threshold odds / exchange rate  [1](https://www.bmj.com/content/352/bmj.i6)[2](https://pmc.ncbi.nlm.nih.gov/articles/PMC6777022/)

  # Standard net benefit: TP/N - FP/N * w  [2](https://pmc.ncbi.nlm.nih.gov/articles/PMC6777022/)[1](https://www.bmj.com/content/352/bmj.i6)
  net_benefit <- (TP / N) - (FP / N) * w

  # Reference strategies for decision curves
  prevalence <- sum(data$outcome_actual == as.character(positive_level), na.rm = TRUE) / N

  # Treat none: NB = 0 by definition (no positives, no false positives) [2](https://pmc.ncbi.nlm.nih.gov/articles/PMC6777022/)[1](https://www.bmj.com/content/352/bmj.i6)
  nb_treat_none <- 0

  # Treat all: TP/N = prevalence; FP/N = (1 - prevalence) [2](https://pmc.ncbi.nlm.nih.gov/articles/PMC6777022/)[1](https://www.bmj.com/content/352/bmj.i6)
  nb_treat_all <- prevalence - (1 - prevalence) * w

  cat("\nThreshold (pt) =", pt,
      "\nNet benefit (model)     =", net_benefit,
      "\nNet benefit (treat all) =", nb_treat_all,
      "\nNet benefit (treat none)=", nb_treat_none,
      "\n")

  return(list(
    confusion_matrix = cm,
    epi_tests = epi,
    net_benefit = net_benefit,
    net_benefit_treat_all = nb_treat_all,
    net_benefit_treat_none = nb_treat_none,
    components = list(TP = TP, FP = FP, N = N, pt = pt, w = w, prevalence = prevalence),
    data = data
  ))
}
```

```python
## Whilever pROC is unavailable, use the following codeblock to produce ROC AUC and curve manually:

ROCdata <- model$data
ROCoutcome <- model$outcome

y <- ROCdata[[ROCoutcome]]
x <- ROCdata$pred

# Remove missing values

ok <- complete.cases(x, y)
x <- x[ok]
y <- y[ok]

# Ensure outcome is binary (0 and 1)

if(is.factor(y)) {
    y <- as.numeric(y == levels(y)[2])
}

# Case and control counts

n_cases    <- sum(y == 1)
n_controls <- sum(y == 0)
cat(sprintf("Cases: %d | Controls: %d\n", n_cases, n_controls))

# Sort by score descending

ord <- order(x, decreasing = TRUE)
x <- x[ord]
y <- y[ord]

# Cumulative counts (true positives and false positives)

tp <- cumsum(y)
fp <- cumsum(1 - y)

P <- sum(y) # Total positives
N <- length(y) - P # Total negatives

tpr <- c(0, tp / P, 1) # True positive rate (sensitivity)
fpr <- c(0, fp / N, 1) # False positive rate (1 - specificity)

# AUC (trapezoidal rule)

auc <- sum(diff(fpr) * (head(tpr, -1) + tail(tpr, -1)) / 2)

# DeLong 95% CI

cases_scores    <- x[y == 1]
controls_scores <- x[y == 0]

V10 <- sapply(cases_scores,    function(xi) mean(ifelse(xi > controls_scores, 1, ifelse(xi == controls_scores, 0.5, 0))))
V01 <- sapply(controls_scores, function(yj) mean(ifelse(cases_scores > yj,    1, ifelse(cases_scores == yj,    0.5, 0))))

var_auc <- var(V10) / n_cases + var(V01) / n_controls
se_auc  <- sqrt(var_auc)

ci_lower <- auc - 1.96 * se_auc
ci_upper <- auc + 1.96 * se_auc

cat(sprintf("AUC: %.4f (95%% CI: %.4f - %.4f)\n", auc, ci_lower, ci_upper))

# Plot ROC curve

plot(fpr, tpr, type = "l", lwd = 2,
     xlab = "False Positive Rate (1 - Specificity)",
     ylab = "True Positive Rate (Sensitivity)",
     main = sprintf("ROC Curve\nAUC = %.4f (95%% CI: %.4f - %.4f)\nCases = %d  |  Controls = %d",
                    auc, ci_lower, ci_upper, n_cases, n_controls))

abline(0, 1, lty = 2, col = "grey")
```
