```r
# This codeblock adds family history of cancers to the dataset

system("bash -c 'unset HTTP_PROXY HTTPS_PROXY http_proxy https_proxy; dx download project-J8PJg1V25y16J0Y3F05pPJBj:/data/raw/questionnaire.tsv'") # Downloads the questionnaire.tsv file from the GOCT project space to the local working directory

questionnaire_cancer_only <- read.table("questionnaire.tsv", header = TRUE, sep = "\t") %>%
    dplyr::select("pid", "father_diag_cancer_1_m", "mother_diag_cancer_1_m", "sibling_diag_cancer_1_m", "father_age_1_1", "mother_age_1_1", "father_age_deceased_1_1", "mother_age_deceased_1_1")

print(paste("Fetched", nrow(questionnaire_cancer_only), "participants from questionnaire data"))

## use grep to highlight any history of breast or prostate cancer in the family history columns

cancer_cols <- c(
  "father_diag_cancer_1_m",
  "mother_diag_cancer_1_m",
  "sibling_diag_cancer_1_m"
)

questionnaire_cancer_only <- questionnaire_cancer_only %>%
  dplyr::mutate(
    FHx_prostate = as.integer(
      dplyr::if_any(
        dplyr::all_of(cancer_cols),
        ~ grepl("(^|\\|)17(\\||$)", dplyr::coalesce(as.character(.x), ""))   # This grep command looks for the code 17 (Prostate cancer), including if it is part of a pipe-separated list of codes in the family history columns
      )
    ),
    FHx_breast = as.integer(
      dplyr::if_any(
        dplyr::all_of(cancer_cols),
        ~ grepl("(^|\\|)4(\\||$)", dplyr::coalesce(as.character(.x), ""))   # This grep command looks for the code 4 (Breast cancer), again including if it is part of a pipe-separated list of codes 
      )
    ),
    FHx_PrCa_BrCa = as.integer(FHx_prostate == 1 | FHx_breast == 1),
    ## add "FHx_adjusted", which censors any negative family history if the parent died below the age of 50, or if parent is currently under 50
    ## in these age variables, -1 and -3 indicate do not know or prefer not to answer, so are captured by the <50 condition
    FHx_prostate_adjusted = case_when(
       FHx_prostate == 1 ~ 1L,
       FHx_prostate == 0 & (father_age_deceased_1_1 < 50 | father_age_1_1 < 50) ~ NA_integer_,
       TRUE ~ 0L
    ),
    FHx_breast_adjusted = case_when(
       FHx_breast == 1 ~ 1L,
       FHx_breast == 0 & (mother_age_deceased_1_1 < 50 | mother_age_1_1 < 50) ~ NA_integer_,
       TRUE ~ 0L
    ),
    FHx_PrCa_BrCa_adjusted = case_when(
      FHx_prostate_adjusted == 1 | FHx_breast_adjusted == 1 ~ 1L,
      is.na(FHx_prostate_adjusted) & is.na(FHx_breast_adjusted) ~ NA_integer_,
      FHx_prostate_adjusted == 0 & FHx_breast_adjusted == 0 ~ 0L,
      TRUE ~ NA_integer_
    )
  )

## Add to main dataframe

all_data <- merge(all_data, questionnaire_cancer_only, by = "pid", all.x = TRUE)

print(paste("Added family history of prostate and breast cancer to main dataset, resulting in", nrow(all_data), "participants"))
```

```r
# Tag participants who already had prostate cancer at a particular "screening date" as "pre_diagnosed". These participants will be filtered out later
# Any death-only PrCa cases will be treated as pre_diagnosed, as we don't have HES or Cancer Registry dates to prove that they were diagnosed after
# the screening date.
# (furthermore, death records are more up-to-date than cancer registry records, so in most cases there is simply a delay in the release of the cancer registry dates)

## Ensure date columns are in date format
date_columns <- c("date_of_birth", "registration_date", "consent_date", "date40", "date45", "date50", "date55", "date60", "date65",
                  "admidate_outpat", "admidate_inpat", "admidate_cancerregistry", "admidate_death", "earliest_PrCa_date",
                  "artificial_screening_date")

all_data_PrCa_GRS[date_columns] <- lapply(all_data_PrCa_GRS[date_columns], as.Date)

## Retrieve the current date
current_date <- as.Date(sub(".*study_ofhs260023_(\\d{8}).*", "\\1", db), format = "%Y%m%d")
print(paste("Current date (from database name):", current_date))

## Override current date because:
## the cancer registry data is only available up to 2023-12-31
## all other case definition data (inpat, outpat, death) is more up-to-date

current_date <- as.Date("2023-12-31")

## Define "pre_diagnosed"

PrCa_prediagnosis <- all_data_PrCa_GRS %>%
    dplyr::mutate(
        pre_diagnosed_registration_date = PrCa == 1 & (                                                                   # pre_diagnosed means any of the following...
            (is.na(admidate_outpat) & is.na(admidate_inpat) & is.na(admidate_cancerregistry) & is.na(admidate_death)) |     # if they have no dates of diagnosis (since we can't be sure when they were diagnosed)
            (!is.na(admidate_outpat) & admidate_outpat < registration_date) |                                               # if they have an outpatient diagnosis date before the screening date    
            (!is.na(admidate_inpat) & admidate_inpat < registration_date) |                                                 # if they have an inpatient diagnosis date before the screening date
            (!is.na(admidate_cancerregistry) & admidate_cancerregistry < registration_date) |                               # if they have a cancer registry diagnosis date before the screening date
            (!is.na(admidate_death) & admidate_death < registration_date)                                                   # if they have a death date before the screening date
        ),
        pre_diagnosed_consent_date = PrCa == 1 & (                                                                   # pre_diagnosed means any of the following...
            (is.na(admidate_outpat) & is.na(admidate_inpat) & is.na(admidate_cancerregistry) & is.na(admidate_death)) |     # if they have no dates of diagnosis (since we can't be sure when they were diagnosed)
            (!is.na(admidate_outpat) & admidate_outpat < consent_date) |                                                    # if they have an outpatient diagnosis date before the screening date    
            (!is.na(admidate_inpat) & admidate_inpat < consent_date) |                                                      # if they have an inpatient diagnosis date before the screening date
            (!is.na(admidate_cancerregistry) & admidate_cancerregistry < consent_date) |                                    # if they have a cancer registry diagnosis date before the screening date
            (!is.na(admidate_death) & admidate_death < consent_date)                                                        # if they have a death date before the screening date
        ),
        pre_diagnosed_asd = PrCa == 1 & (                                                                                       # pre_diagnosed means any of the following...
            (is.na(admidate_outpat) & is.na(admidate_inpat) & is.na(admidate_cancerregistry) & is.na(admidate_death)) |
            (!is.na(admidate_outpat) & admidate_outpat < artificial_screening_date) |                                       # if they have an outpatient diagnosis date before the screening date
            (!is.na(admidate_inpat) & admidate_inpat < artificial_screening_date) |                                         # if they have an inpatient diagnosis date before the screening date
            (!is.na(admidate_cancerregistry) & admidate_cancerregistry < artificial_screening_date) |
            (!is.na(admidate_death) & admidate_death < artificial_screening_date)
        ),        
        pre_diagnosed_date40 = PrCa == 1 & (                                                                                   # pre_diagnosed means any of the following...
            (is.na(admidate_outpat) & is.na(admidate_inpat) & is.na(admidate_cancerregistry) & is.na(admidate_death)) |
            (!is.na(admidate_outpat) & admidate_outpat < date40) |                                                             # if they have an outpatient diagnosis date before the screening date
            (!is.na(admidate_inpat) & admidate_inpat < date40) |                                                               # if they have an inpatient diagnosis date before the screening date
            (!is.na(admidate_cancerregistry) & admidate_cancerregistry < date40) |
            (!is.na(admidate_death) & admidate_death < date40)
        ),
        pre_diagnosed_date45 = PrCa == 1 & (
            (is.na(admidate_outpat) & is.na(admidate_inpat) & is.na(admidate_cancerregistry) & is.na(admidate_death)) |
            (!is.na(admidate_outpat) & admidate_outpat < date45) |
            (!is.na(admidate_inpat) & admidate_inpat < date45) |
            (!is.na(admidate_cancerregistry) & admidate_cancerregistry < date45) |
            (!is.na(admidate_death) & admidate_death < date45)
        ),
        pre_diagnosed_date50 = PrCa == 1 & (
            (is.na(admidate_outpat) & is.na(admidate_inpat) & is.na(admidate_cancerregistry) & is.na(admidate_death)) |
            (!is.na(admidate_outpat) & admidate_outpat < date50) |
            (!is.na(admidate_inpat) & admidate_inpat < date50) |
            (!is.na(admidate_cancerregistry) & admidate_cancerregistry < date50) |
            (!is.na(admidate_death) & admidate_death < date50)
        ),
        pre_diagnosed_date55 = PrCa == 1 & (
            (is.na(admidate_outpat) & is.na(admidate_inpat) & is.na(admidate_cancerregistry) & is.na(admidate_death)) |
            (!is.na(admidate_outpat) & admidate_outpat < date55) |
            (!is.na(admidate_inpat) & admidate_inpat < date55) |
            (!is.na(admidate_cancerregistry) & admidate_cancerregistry < date55) |
            (!is.na(admidate_death) & admidate_death < date55)
        ),
        pre_diagnosed_date60 = PrCa == 1 & (
            (is.na(admidate_outpat) & is.na(admidate_inpat) & is.na(admidate_cancerregistry) & is.na(admidate_death)) |
            (!is.na(admidate_outpat) & admidate_outpat < date60) |
            (!is.na(admidate_inpat) & admidate_inpat < date60) |
            (!is.na(admidate_cancerregistry) & admidate_cancerregistry < date60) |
            (!is.na(admidate_death) & admidate_death < date60)
        ),
        pre_diagnosed_date65 = PrCa == 1 & (
            (is.na(admidate_outpat) & is.na(admidate_inpat) & is.na(admidate_cancerregistry) & is.na(admidate_death)) |
            (!is.na(admidate_outpat) & admidate_outpat < date65) |
            (!is.na(admidate_inpat) & admidate_inpat < date65) |
            (!is.na(admidate_cancerregistry) & admidate_cancerregistry < date65) |
            (!is.na(admidate_death) & admidate_death < date65)
        ),
       earliest_PrCa_date = pmin(admidate_outpat, admidate_inpat, admidate_cancerregistry, admidate_death, na.rm = TRUE),
       earliest_PrCa_date = dplyr::if_else(is.infinite(as.numeric(earliest_PrCa_date)),as.Date(NA), as.Date(earliest_PrCa_date))
    )

## read in dates of death

death_registry <- DBI::dbGetQuery(sc, sprintf("SELECT * FROM %s.nhse_engwal_deaths", db))
write.csv(death_registry,"death_registry.csv", row.names = FALSE)

death_dates <- death_registry %>%
  dplyr::select(pid, reg_date_of_death) %>%
  dplyr::mutate(reg_date_of_death = as.Date(reg_date_of_death)) %>%
  dplyr::group_by(pid) %>%
  dplyr::summarise(
    reg_date_of_death = if (all(is.na(reg_date_of_death))) as.Date(NA) else min(reg_date_of_death, na.rm = TRUE),
    .groups = "drop"
  )

## Tag participants who had a prostate cancer record in inpatient and/or outpatient data but not in cancer registry or death records as "HES-only" (these will be filtered out later)
# also tag participants with prostate cancer record in death records but not in inpatient, outpatient, or cancer registry data as "death-only" (these will also be filtered out later)
# also add "current age" and "old_enough" flag - if current age is less than the screening age + 2, 5, or 10 years, then old enough = FALSE
# also add "died_within_2yrs", "died_within_5yrs", and "died_within_10yrs" flags - if they died within 2, 5, or 10 years of the screening date, then these flags = TRUE

PrCa_prediagnosis <- PrCa_prediagnosis %>%
    dplyr::mutate(
        HES_only = PrCa == 1 & ((is.na(admidate_cancerregistry) & is.na(admidate_death)) & (!is.na(admidate_outpat) | !is.na(admidate_inpat))),
        death_only = PrCa == 1 & (is.na(admidate_outpat) & is.na(admidate_inpat) & is.na(admidate_cancerregistry) & !is.na(admidate_death)),
        current_age = as.numeric(difftime(current_date, date_of_birth, units = "days")) / 365.25,
        over_40 = current_age >= 40,
        over_42 = current_age >= 42,
        over_45 = current_age >= 45,
        over_47 = current_age >= 47,
        over_50 = current_age >= 50,
        over_52 = current_age >= 52,
        over_55 = current_age >= 55,                # These dates cover all the screening dates + 2, 5, and 10 years
        over_57 = current_age >= 57,
        over_60 = current_age >= 60,
        over_62 = current_age >= 62,
        over_65 = current_age >= 65,
        over_67 = current_age >= 67,
        over_70 = current_age >= 70,
        over_75 = current_age >= 75,
        ### Add extras for artificial screening date (asd)
        over_20_at_asd = age_asd >= 20,
        over_30_at_asd = age_asd >= 30,
        over_40_at_asd = age_asd >= 40
    ) %>%
    dplyr::left_join(death_dates, by = "pid") %>%
    dplyr::mutate(
        died_by_40 = !is.na(reg_date_of_death) & (reg_date_of_death <= date_of_birth + lubridate::years(40)),
        died_by_42 = !is.na(reg_date_of_death) & (reg_date_of_death <= date_of_birth + lubridate::years(42)),
        died_by_45 = !is.na(reg_date_of_death) & (reg_date_of_death <= date_of_birth + lubridate::years(45)),
        died_by_47 = !is.na(reg_date_of_death) & (reg_date_of_death <= date_of_birth + lubridate::years(47)),
        died_by_50 = !is.na(reg_date_of_death) & (reg_date_of_death <= date_of_birth + lubridate::years(50)),
        died_by_52 = !is.na(reg_date_of_death) & (reg_date_of_death <= date_of_birth + lubridate::years(52)),
        died_by_55 = !is.na(reg_date_of_death) & (reg_date_of_death <= date_of_birth + lubridate::years(55)),
        died_by_57 = !is.na(reg_date_of_death) & (reg_date_of_death <= date_of_birth + lubridate::years(57)),
        died_by_60 = !is.na(reg_date_of_death) & (reg_date_of_death <= date_of_birth + lubridate::years(60)),
        died_by_62 = !is.na(reg_date_of_death) & (reg_date_of_death <= date_of_birth + lubridate::years(62)),
        died_by_65 = !is.na(reg_date_of_death) & (reg_date_of_death <= date_of_birth + lubridate::years(65)),
        died_by_67 = !is.na(reg_date_of_death) & (reg_date_of_death <= date_of_birth + lubridate::years(67)),
        died_by_70 = !is.na(reg_date_of_death) & (reg_date_of_death <= date_of_birth + lubridate::years(70)),
        died_by_75 = !is.na(reg_date_of_death) & (reg_date_of_death <= date_of_birth + lubridate::years(75)),
        # Add extra for artificial screening date (asd)
        died_by_asd = !is.na(reg_date_of_death) & (reg_date_of_death <= artificial_screening_date),
        died_by_asd_plus_2yrs = !is.na(reg_date_of_death) & (reg_date_of_death <= artificial_screening_date + lubridate::years(2)),
        died_by_asd_plus_5yrs = !is.na(reg_date_of_death) & (reg_date_of_death <= artificial_screening_date + lubridate::years(5)),
        died_by_asd_plus_10yrs = !is.na(reg_date_of_death) & (reg_date_of_death <= artificial_screening_date + lubridate::years(10))
    )

write.csv(PrCa_prediagnosis, "PrCa_prediagnosis.csv", row.names = FALSE)
```

```r
# Create dataframes ready for logistic regression analyses

# Set the screening date variable here, to any of the following:
#    - artificial_screening_date
#    - registration_date
#    - date40
#    - date45
#    - date50
#    - date55
#    - date60
#    - date65

screening_date <- "artificial_screening_date" 

## Create predhorizon dataframes

library(lubridate)

PrCa_predhorizon <- PrCa_prediagnosis_males %>%
    dplyr::mutate(
        PrCa_post_assessment  = as.integer(
            !is.na(earliest_PrCa_date) & !is.na(.data[[screening_date]]) &
                earliest_PrCa_date >= .data[[screening_date]]
        ),
        PrCa_2yrs = as.integer(
            !is.na(earliest_PrCa_date) & !is.na(.data[[screening_date]]) &
                earliest_PrCa_date >= .data[[screening_date]] &
                earliest_PrCa_date <= .data[[screening_date]] %m+% lubridate::years(2)
        ),
        PrCa_5yrs = as.integer(
            !is.na(earliest_PrCa_date) & !is.na(.data[[screening_date]]) &
                earliest_PrCa_date >= .data[[screening_date]] &
                earliest_PrCa_date <= .data[[screening_date]] %m+% lubridate::years(5)
        ),
        PrCa_10yrs = as.integer(
            !is.na(earliest_PrCa_date) & !is.na(.data[[screening_date]]) &
                earliest_PrCa_date >= .data[[screening_date]] &
                earliest_PrCa_date <= .data[[screening_date]] %m+% lubridate::years(10)
        )
    )

print(paste("After filtering to males only, there are", nrow(PrCa_predhorizon), "participants, down from", nrow(PrCa_prediagnosis), "participants"))


## Create separate dataframes for ethnicity groups and filter out HES only cases, death only cases, pre-diagnosed cases, participants without screening dates, participants without linked genetic data, and participants who are not old enough for/died before the screening date

if (screening_date == "artificial_screening_date") {
    PrCa_predhorizon_allmen <- PrCa_predhorizon %>%
        dplyr::filter(
            HES_only == FALSE & death_only == FALSE & pre_diagnosed_asd == FALSE & !is.na(.data[[screening_date]]) & !is.na(ContiMultiethnicGRS) & over_20_at_asd == TRUE & died_by_asd == FALSE
            # Note - ContiMultiethnicGRS used as genetic-data availability marker
        )
} else if (screening_date == "date40") {
    PrCa_predhorizon_allmen <- PrCa_predhorizon %>%
        dplyr::filter(
            HES_only == FALSE & death_only == FALSE & pre_diagnosed_date40 == FALSE & !is.na(.data[[screening_date]]) & !is.na(ContiMultiethnicGRS) & over_40 == TRUE & died_by_40 == FALSE
        )
} else if (screening_date == "date45") {
    PrCa_predhorizon_allmen <- PrCa_predhorizon %>%
        dplyr::filter(
            HES_only == FALSE & death_only == FALSE & pre_diagnosed_date45 == FALSE & !is.na(.data[[screening_date]]) & !is.na(ContiMultiethnicGRS) & over_45 == TRUE & died_by_45 == FALSE
        )
} else if (screening_date == "date50") {
    PrCa_predhorizon_allmen <- PrCa_predhorizon %>%
        dplyr::filter(
            HES_only == FALSE & death_only == FALSE & pre_diagnosed_date50 == FALSE & !is.na(.data[[screening_date]]) & !is.na(ContiMultiethnicGRS) & over_50 == TRUE & died_by_50 == FALSE
        )
} else if (screening_date == "date55") {
    PrCa_predhorizon_allmen <- PrCa_predhorizon %>%
        dplyr::filter(
            HES_only == FALSE & death_only == FALSE & pre_diagnosed_date55 == FALSE & !is.na(.data[[screening_date]]) & !is.na(ContiMultiethnicGRS) & over_55 == TRUE & died_by_55 == FALSE
        )
} else if (screening_date == "date60") {
    PrCa_predhorizon_allmen <- PrCa_predhorizon %>%
        dplyr::filter(
            HES_only == FALSE & death_only == FALSE & pre_diagnosed_date60 == FALSE & !is.na(.data[[screening_date]]) & !is.na(ContiMultiethnicGRS) & over_60 == TRUE & died_by_60 == FALSE
        )
} else if (screening_date == "date65") {
    PrCa_predhorizon_allmen <- PrCa_predhorizon %>%
        dplyr::filter(
            HES_only == FALSE & death_only == FALSE & pre_diagnosed_date65 == FALSE & !is.na(.data[[screening_date]]) & !is.na(ContiMultiethnicGRS) & over_65 == TRUE & died_by_65 == FALSE
        )
}

PrCa_predhorizon <- PrCa_predhorizon %>%
    dplyr::mutate(
        died_by_screening_date_flag = case_when(
            screening_date == "artificial_screening_date" ~ died_by_asd,
            screening_date == "date40" ~ died_by_40,
            screening_date == "date45" ~ died_by_45,
            screening_date == "date50" ~ died_by_50,
            screening_date == "date55" ~ died_by_55,
            screening_date == "date60" ~ died_by_60,
            screening_date == "date65" ~ died_by_65,
            TRUE ~ NA
        )
    )

print(paste("Prior to filtering, there are", nrow(PrCa_predhorizon), "male participants in the predhorizon dataframe, of which", sum(PrCa_predhorizon$PrCa == 1, na.rm = TRUE), "are PrCa cases."))
print(paste("After filtering out", nrow(PrCa_predhorizon %>% dplyr::filter(HES_only == TRUE & PrCa == 1)), "HES-only cases, ", nrow(PrCa_predhorizon %>% dplyr::filter(death_only == TRUE & PrCa == 1)), "death-only cases, ", nrow(PrCa_predhorizon %>% dplyr::filter(pre_diagnosed_asd == TRUE & PrCa == 1)), "pre-diagnosed cases, ", nrow(PrCa_predhorizon %>% dplyr::filter(is.na(.data[[screening_date]]))), "participants without screening dates, ", nrow(PrCa_predhorizon %>% dplyr::filter(is.na(ContiMultiethnicGRS))), "participants without linked genetic data, ", nrow(PrCa_predhorizon %>% dplyr::filter(over_20_at_asd == FALSE)), "participants who are too young, or ", nrow(PrCa_predhorizon %>% dplyr::filter(died_by_screening_date_flag == TRUE)), "participants who died before the screening date, there are", nrow(PrCa_predhorizon_allmen), "men remaining for analysis at screening date", screening_date, "of which", sum(PrCa_predhorizon_allmen$PrCa_post_assessment == 1, na.rm = TRUE), "are incident PrCa cases."))

## Subset to population groups

PrCa_predhorizon_whitemen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(
        ethnicity_group_narrow == "White"
    )

PrCa_predhorizon_blackmen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(
        ethnicity_group_narrow == "Black"
    )

PrCa_predhorizon_mixedmen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(
        ethnicity_group_narrow == "Mixed White and Black"
    )

PrCa_predhorizon_southasianmen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(
        ethnicity_group_narrow == "South Asian"
    )

PrCa_predhorizon_chinesemen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(
        ethnicity_group_narrow == "Chinese"
    )

PrCa_predhorizon_arabmen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(
        ethnicity_group_narrow == "Arab"
    )

PrCa_predhorizon_othermen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(
        ethnicity_group_narrow == "Other"
    )

PrCa_predhorizon_mixed_other <- PrCa_predhorizon_allmen %>%
    dplyr::filter(
        ethnicity_group_narrow %in% c("Mixed White and Black", "Mixed White and Asian", "Other")
    )

# Create separate dataframes for genetic similarity groups

PrCa_predhorizon_EURmen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(genetic_similarity_group == "EUR")

PrCa_predhorizon_AFRmen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(genetic_similarity_group == "AFR")

PrCa_predhorizon_SASmen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(genetic_similarity_group == "SAS")

PrCa_predhorizon_EASmen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(genetic_similarity_group == "EAS")

PrCa_predhorizon_SEAmen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(genetic_similarity_group == "SEA")

PrCa_predhorizon_CASmen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(genetic_similarity_group == "CAS")

PrCa_predhorizon_AMRmen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(genetic_similarity_group == "AMR")

PrCa_predhorizon_MIDmen <- PrCa_predhorizon_allmen %>%
    dplyr::filter(genetic_similarity_group == "MID")


# Write all to .csv to take a look

write.csv(PrCa_predhorizon_allmen, "PrCa_predhorizon_allmen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_whitemen, "PrCa_predhorizon_whitemen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_blackmen, "PrCa_predhorizon_blackmen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_mixedmen, "PrCa_predhorizon_mixedmen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_southasianmen, "PrCa_predhorizon_southasianmen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_chinesemen, "PrCa_predhorizon_chinesemen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_arabmen, "PrCa_predhorizon_arabmen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_othermen, "PrCa_predhorizon_othermen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_mixed_other, "PrCa_predhorizon_mixed_other.csv", row.names = FALSE)

write.csv(PrCa_predhorizon_EURmen, "PrCa_predhorizon_EURmen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_AFRmen, "PrCa_predhorizon_AFRmen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_SASmen, "PrCa_predhorizon_SASmen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_EASmen, "PrCa_predhorizon_EASmen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_SEAmen, "PrCa_predhorizon_SEAmen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_CASmen, "PrCa_predhorizon_CASmen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_AMRmen, "PrCa_predhorizon_AMRmen.csv", row.names = FALSE)
write.csv(PrCa_predhorizon_MIDmen, "PrCa_predhorizon_MIDmen.csv", row.names = FALSE)
```
